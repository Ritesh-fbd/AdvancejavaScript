```python
#Slicing arrays
#Slicing in python means taking elements from one given index to another given index.

#We pass slice instead of index like this: [start:end].

#We can also define the step, like this: [start:end:step].

#If we don't pass start its considered 0

#If we don't pass end its considered length of array in that dimension

#If we don't pass step its considered 1
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5])
```

```python
def linear_search(arr, target):
    for i in range(len(arr)):  # Go through each item in the list
        if arr[i] == target:   # If the current item is the target
            return i           # Return the position (index)
    return -1                  # If not found, return -1

# Example usage
arr = [5, 8, 12, 20, 35]
target = 12

result = linear_search(arr, target)

if result != -1:
    print(f"Found at index {result}")
else:
    print("Not found")
```

```python
# Linear search without using a function
array = [10, 25, 30, 45, 50, 65, 80]
target = int(input("Enter the value to search for: "))

found = False
for i in range(len(array)):
    if array[i] == target:
        print(f"Value {target} found at index {i}")
        found = True
        break

if not found:
    print(f"Value {target} not found in the array")
```

```python
import numpy as np

# Create two arrays
array1 = np.array([[1, 2], [3, 4]])
array2 = np.array([[5, 6], [7, 8]])

# Joining arrays using concatenate
joined_array = np.concatenate((array1, array2), axis=0)  # Join along rows (axis=0)
print("Joined array (concatenate along rows):\n", joined_array)

# Joining arrays using vstack (Vertical stacking)
vstack_array = np.vstack((array1, array2))
print("\nVertically stacked array:\n", vstack_array)

# Joining arrays using hstack (Horizontal stacking)
hstack_array = np.hstack((array1, array2))
print("\nHorizontally stacked array:\n", hstack_array)
```

```python
# Splitting arrays using split
split_array = np.split(joined_array, 2, axis=0)  # Split into 2 sub-arrays along rows
print("\nSplit arrays (split along rows):")
for arr in split_array:
    print(arr)

# Splitting arrays using array_split (unequal splitting)
unequal_split = np.array_split(hstack_array, 3, axis=1)  # Split into 3 parts along columns
print("\nUnequal split arrays (array_split along columns):")
for arr in unequal_split:
    print(arr)
```

```python
import numpy as np

# Create a 2D array (matrix)
array = np.array([[1, 2, 3], 
                  [4, 5, 6]])

# Transpose the array using .T (shorthand)
transposed_array_T = array.T
print("Transposed array using .T:\n", transposed_array_T)

# Transpose the array using transpose() function
transposed_array_func = np.transpose(array)
print("\nTransposed array using transpose():\n", transposed_array_func)
```

```python

In NumPy, data types (also known as dtypes) define the type of elements that an array can hold.
Each array in NumPy has a single data type for all of its elements. Some of the most common data types in NumPy include:

int: Integer data type (e.g., int32, int64)
float: Floating point data type (e.g., float32, float64)
complex: Complex numbers (e.g., complex64, complex128)
bool: Boolean data type (True or False)
str: Strings
object: Python objects
```

```python
import numpy as np

# Integer array
int_array = np.array([1, 2, 3, 4], dtype=np.int32)
print("Integer Array:\n", int_array)
print("Data Type:", int_array.dtype)

# Float array
float_array = np.array([1.5, 2.5, 3.5, 4.5], dtype=np.float64)
print("\nFloat Array:\n", float_array)
print("Data Type:", float_array.dtype)

# Complex array
complex_array = np.array([1+2j, 3+4j], dtype=np.complex128)
print("\nComplex Array:\n", complex_array)
print("Data Type:", complex_array.dtype)

# Boolean array
bool_array = np.array([True, False, True], dtype=np.bool_)
print("\nBoolean Array:\n", bool_array)
print("Data Type:", bool_array.dtype)

# String array
string_array = np.array(['apple', 'banana', 'cherry'], dtype=np.str_)
print("\nString Array:\n", string_array)
print("Data Type:", string_array.dtype)

# Object array (holding Python objects)
object_array = np.array([{'name': 'Alice'}, {'name': 'Bob'}], dtype=np.object_)
print("\nObject Array:\n", object_array)
print("Data Type:", object_array.dtype)
```

```python
# Convert integer array to float
converted_array = int_array.astype(np.float64)
print("\nConverted Array (int to float):\n", converted_array)
print("New Data Type:", converted_array.dtype)
```

```python
import numpy as np

# Create two arrays
array1 = np.array([1, 2, 3, 4])
array2 = np.array([5, 6, 7, 8])

# Element-wise addition
addition = array1 + array2
print("Addition:\n", addition)

# Element-wise subtraction
subtraction = array2 - array1
print("\nSubtraction:\n", subtraction)

# Element-wise multiplication
multiplication = array1 * array2
print("\nMultiplication:\n", multiplication)

# Element-wise division
division = array2 / array1
print("\nDivision:\n", division)

# Element-wise exponentiation
exponentiation = array1 ** 2
print("\nExponentiation (square of array1):\n", exponentiation)

# Element-wise modulus (remainder of division)
modulus = array2 % array1
print("\nModulus (remainder of array2 / array1):\n", modulus)

# Element-wise floor division
floor_division = array2 // array1
print("\nFloor Division (array2 // array1):\n", floor_division)
```

```python
#Broadcasting:
#In NumPy, broadcasting allows you to perform arithmetic between arrays of different shapes, as long as the shapes are compatible.


# Scalar and array broadcasting
scalar_addition = array1 + 10
print("\nScalar Addition (array1 + 10):\n", scalar_addition)
```
