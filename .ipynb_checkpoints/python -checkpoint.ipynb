```python
#Python program to swap elements in the list
inputlist=input("please enter list elements").split()
a=int(input("enter 1st element index to swap"))
b=int(input("enter 2nd element index to swap"))
length=len(inputlist)
if(a>=length or b>=length):
    print("Indexes are out of boundry of Input list")
else:
    print("Original List: ",inputlist)
    temp=inputlist[a]
    inputlist[a]=inputlist[b]
    inputlist[b]=temp
    print("List after swapping: ",inputlist)
```

```python
def is_perfect(num):
 return sum(i for i in range(1, num) if num % i == 0) == num
perfect_numbers = [num for num in range(1, 1001) if is_perfect(num)]
print("Perfect numbers:", perfect_numbers)
```
