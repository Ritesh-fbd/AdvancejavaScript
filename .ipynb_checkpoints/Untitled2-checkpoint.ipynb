```python
m1=float(input("Enter the marks of 1st subject:"))
m2=float(input("Enter the marks of 2nd subject:"))
m3=float(input("Enter the marks of 3rd subject:"))
m4=float(input("Enter the marks of 4th subject:"))
m5=float(input("Enter the marks of 5th subject:"))
avr=(m1+m2+m3+m4+m5)/5
if(avr>90):
    print("Grade A")
elif(avr>80 and avr<=90):
    print("Grade B")
elif(avr>70 and avr<=80):
    print("Grade C")
elif(avr>60 and avr<=70):
    print("Grade D")
elif(avr>50and avr<=60):
    print("Grade E")
else:
    print("Fail")
```

```python
bs=int(input("Enter the basic salary:"))
if(bs<20000):
    da=(.1*bs)
else:
    da=(.15*bs)
if(bs<=10000):
    ta=(.05*bs)
elif(bs>10000 and bs<=20000):
    ta=(.1*bs)
else:
    ta=(.2*bs)
gs= (bs*.45)+da+ta+bs
print(gs)
if(bs<=20000):
    tds=(.1*gs)
else:
    tds=(.15*gs)
ts=gs-tds
print(ts)
```

```python
basic_salary = float(input("Enter the basic salary: "))
da = 0.45 * basic_salary
if basic_salary <= 20000:
    hra = 0.10 * basic_salary
else:
    hra = 0.15 * basic_salary
if basic_salary <= 10000:
    ta = 0.05 * basic_salary
elif basic_salary <= 20000:
    ta = 0.10 * basic_salary
else:
    ta = 0.20 * basic_salary
gross_salary = basic_salary + da + hra + ta
if gross_salary <= 20000:
    tds = 0.10 * gross_salary
else:
    tds = 0.15 * gross_salary
net_salary = gross_salary - tds
print("\nSalary Details:")
print(f"Basic Salary: {basic_salary:.2f}")
print(f"DA: {da:.2f}")
print(f"HRA: {hra:.2f}")
print(f"TA: {ta:.2f}")
print(f"Gross Salary: {gross_salary:.2f}")
print(f"TDS: {tds:.2f}")
print(f"Net Salary: {net_salary:.2f}")
```
