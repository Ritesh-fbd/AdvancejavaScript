```python
def linear_search(arr,target):
    for i in range(len(arr)):
        if arr[i]==target: 
            return i
    return -1
arr=[5,8,9,12,20,35] 
target=12 
r=linear_search(arr,target)
if r!=-1:
    print(f"found at index {r}")
else: 
    print("notfound")
```
